# Pooled OLS Regression Model

In this panel dataset, we have collected data for 38 US states, or "units", over a period of time from 2015 to 2021. This dataset is a "fixed and balanced panel" since each state is tracked over the same number of time periods, and the same set of states are tracked in each time period.

For each unit, we have included 29 explanatory variables, which all change across time periods. We wish to study the impact of these measured variables on the response variable, which is the total number of bee colonies in a state. Thus, we have a three-dimensional dataset in which each row represents a unique unit, each column contains the data from one of the explanatory variables for that unit, and the "z-axis" contains the sequence of time perios over which the unit has been tracked.

Our goal is to choose and fit a regression model that is suitable for this panel dataset. Our regression model should allow us to express yearly bee colony numbers experienced by state $i$ at time period (i.e., year) $t$ as some function $f(\cdot)$ of the explanatory variables in state $i$ during time period $t$. This can be written in the following form

$$y_{t}^{i}=f(x_{t}^{i})+\epsilon_{t}^{i}$$

where $x_{t}^{i}$ is our $(n+1) \times 1$ feature vector (i.e., $n=29$ plus 1 constant), and $\epsilon^{i}_{t}$ is the residual error of regression with a certain mean and variance, and it captures the variance that the model isn't able to "explain." 

With the Pooled OLS regresion model, the equation above can be rewritten as a linear equation. For observational units $i = 1, \ldots, m=39$, we have

$$\mathbf{y}^i = \mathbf{X}^i\beta + \mathbf{\epsilon}^i$$

where $\mathbf{y}^i$ is a $T \times 1$ vector of response variables, $\mathbf{\epsilon}^{i}$ is a $T \times 1$ vector of error terms, $\mathbf{X}^i$ is a $T \times (n+1)$ matrix written as 

$$\mathbf{X}^{i}=\left[\begin{array}{c}
(x_{1}^{i})^{\text{T}}\\
(x_{2}^{i})^{\text{T}}\\
\vdots\\
(x_{T}^{i})^{\text{T}}
\end{array}\right]$$

and $\beta$ is the $(n+1) \times 1$ population parameter we want to estimate. So, we have the same $\beta$ for all $t = 1,\ldots,T=7$. The objective of training the model on the panel data dataset is to find the fitted vector of coefficients $\hat{\beta}$, and the pooled OLS estimator is given by

$$\begin{array}{lcl}
\hat{\beta} & = & \left({\displaystyle \sum_{i=1}^{m}}\left[\mathbf{X}^{i}\right]^{\text{T}}\mathbf{X}^{i}\right)^{-1}\left({\displaystyle \sum_{i=1}^{m}}\left[\mathbf{X}^{i}\right]^{\text{T}}\mathbf{y}^{i}\right)\\
 & = & \left({\displaystyle \sum_{i=1}^{m}}{\displaystyle \sum_{t=1}^{T}}\left[x_{t}^{i}\right]^{\text{T}}x_{t}^{i}\right)^{-1}\left({\displaystyle \sum_{i=1}^{m}}{\displaystyle \sum_{t=1}^{T}}\left[x_{t}^{i}\right]^{\text{T}}y_{t}^{i}\right)
\end{array}$$

The pooled OLS regression model is often a good starting point and reference for other models related to panel data, such as the "Fixed Effects Regression Model" or the "Random Effects Regression Model."

```python
rmse = []
mae = []
mape = []

for t in ['2017-01-01','2018-01-01','2019-01-01','2020-01-01','2021-01-01']:
    # Split the datasets into training and test sets
    df_train = df_b[df_b['year'] < t].copy()
    df_train.sort_values(by=['state','year'], axis=0, ascending=True, inplace=True)
    df_train.set_index(['state','year'], inplace=True)
    
    df_test = df_b[df_b['year']==t].copy()
    df_test.sort_values(by=['state','year'], axis=0, ascending=True, inplace=True)
    df_test.set_index(['state','year'], inplace=True)
    
    # X matrix from the panel data
    # Add a placeholder for the regression intercept
    exog = sm.tools.tools.add_constant(df_train[df_train.columns[1:]])
    # y vector from the panel data
    endog = df_train['colonies_inventory']
    # Build the Pooled OLS regression model
    pols_mod = PooledOLS(endog, exog)
    # Train the model on the (y,X) datasets and retrieve the training results
    pols_results = pols_mod.fit(cov_type='clustered', cluster_entity=True)
    
    # Using the exogenous variables in the testing dataset, make predictions for all 38 states for the following year
    exog = sm.tools.tools.add_constant(df_test[df_test.columns[1:]])
    y_pred = pols_results.predict(exog)
    
    # Combine the true values with the predicted values
    y_true = df_test[['colonies_inventory']]
    df_y = y_true.merge(y_pred, how='inner', left_index=True, right_index=True)
    
    # Calculate the performance metrics
    rmse.appen(mean_squared_error(df_y['colonies_inventory'].to_numpy(), df_y['predictions'].to_numpy(), squared=False))
    mae.append(mean_absolute_error(df_y['colonies_inventory'].to_numpy(), df_y['predictions'].to_numpy()))
    mape.append(mean_absolute_percentage_error(df_y['colonies_inventory'].to_numpy(), df_y['predictions'].to_numpy()))
```
