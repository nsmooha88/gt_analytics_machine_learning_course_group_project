```python
import pandas as pd
```

```python
main_df = pd.read_csv('Quarterly_Added & Replaced.csv')
```

```python
main_df = main_df[['Year','Period','State']]
```

```python
for doc in ['Quarterly_Affected by Disease_PCT.csv','Quarterly_Affected by Mites_PCT.csv','Quarterly_Affected by Other Causes_PCT.csv','Quarterly_Affected by Pesticides_PCT.csv','Quarterly_Affected by Pests_PCT.csv','Quarterly_Affected by Unknown Causes_PCT.csv','Quarterly_Inventory, Max.csv','Quarterly_Loss, Colony Collapse Disorder.csv','Quarterly_Loss, Deadout.csv','Quarterly_Renovated_PCT.csv','Quarterly_Renovated.csv','Quaterly_Loss, Deadout_PCT.csv']:
    df = pd.read_csv(doc)
    df = df[['Year','Period','State']]
    main_df = pd.concat([main_df,df], axis=0, ignore_index=True)
```

```python
main_df.drop_duplicates(keep='first', inplace=True)
```

```python
for doc in ['Quarterly_Added & Replaced.csv','Quarterly_Affected by Disease_PCT.csv','Quarterly_Affected by Mites_PCT.csv','Quarterly_Affected by Other Causes_PCT.csv','Quarterly_Affected by Pesticides_PCT.csv','Quarterly_Affected by Pests_PCT.csv','Quarterly_Affected by Unknown Causes_PCT.csv','Quarterly_Inventory, Max.csv','Quarterly_Loss, Colony Collapse Disorder.csv','Quarterly_Loss, Deadout.csv','Quarterly_Renovated_PCT.csv','Quarterly_Renovated.csv','Quaterly_Loss, Deadout_PCT.csv']:
    df = pd.read_csv(doc)
    main_df = main_df.merge(df, how='left', on=['Year','Period','State'])
```

```python
main_df.to_csv(path_or_buf='bee_colony_data.csv', index='False')
```
