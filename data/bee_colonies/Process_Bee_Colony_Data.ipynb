```python
import openpyxl
import pandas as pd
import duckdb
```

```python
mydf = pd.read_csv('final_data.csv')
```

```python
mydf.head()
```

```python
df = duckdb.query("""
SELECT 
    year,
    state,
    AVG(colonies_inventory_max) AS colonies_inventory_max,
    AVG(colonies_added_replaced) AS colonies_added_replaced,
    AVG(colonies_renovated) AS colonies_renovated,
    AVG(colonies_loss_deadout) AS colonies_loss_deadout,
    AVG(colonies_loss_ccd) AS colonies_loss_ccd,
    AVG(colonies_renovated_pct) AS colonies_renovated_pct,
    AVG(colonies_loss_deadout_pct) AS colonies_loss_deadout_pct,
    AVG(affected_by_disease_pct) AS colonies_disease_pct,
    AVG(affected_by_varroa_mites_pct) AS colonies_varroa_mites_pct,
    AVG(affected_by_other_causes_pct) AS colonies_other_causes_pct,
    AVG(affected_by_pesticides_pct) AS colonies_pesticides_pct,
    AVG(affected_by_other_pests_pct) AS colonies_other_pests_pct,
    AVG(affected_by_unknown_causes_pct) AS colonies_unknown_causes_pct
FROM 
    mydf
GROUP BY
    year, state
""").to_df()
```

```python
df.to_csv('annual_state_usda.csv', header='True')
```
